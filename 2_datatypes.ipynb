```python
# Why is it important to learn about data's types:
    # we wont be able to perform operations on data until we get
    # to know its type.

# What are the different types of data we have:
    # Numeric data
        # Integer:
            # A number without decimal point
            # Including 0
            # +ve or -ve
            # Immutable datatype [Can be modified after its creation]
            # Ex: 5, -10, 0, 12, 450, 567994, -263735

        # Float:
            # A number with decimal point
            # +ve or -ve
            # It is also Immutable
            # Ex: 5.23, 0.000, 6.0001, -0.234, -567.12. -40.00

        # Complex:
            # A number made up by combining:
                            # It is also Immutable
                            # Real Number:
                                # All Integers and Floats
                                # Ex: 4, 5.5, 0.0, 0, 12.9, -34.56, -0.001

                            # Imaginary Number:
                                # A number siffixed by "j" [in math its "i"]
                                # Ex: 5j, 90j, -12j, -3.4j, 7.8j

    # Boolean data
        # True => 1
        # False => 0
        # these are also immutables

    # text data
        # String:
            # A collection of charecters [A-Z, a-z, 0-9, sp_chars], enclosed by quotes.
            # It is also Immutable
            # Ex: 'Hello', 'Admin@123', '123455', '@#$%', 'SDFGH', 'ertyu'

            # What are the quotes used to form a string:
                # single quotes ''
                # double quotes ""
                # tripple single quotes ''' '''
                # tripple double quotes """ """
            
            # Ex: 'Hello "brother"', "Hello 'brother'"

            # NOTE: When String is very Big [Paragraph] then create it using tripple quotes
            # Ex: '''Hello Brother..! This is 'Python' session. We are learning """the worlds most famous and highly used programming language""".'''
```

```python
# Integer:
x = 5
print(x)
print(type(x))  # int

# Float:
y = 3.3
print(y)
print(type(y))  #float

# Complex:
z = 45+56j
print(z)
print(type(z)) #complex
print(z.real)
print(z.imag)
```

```python
a = -45
print(a)
print(type(a))

b = -23.45
print(b)
print(type(b))

c = -12-0.2j
print(c)
print(type(c))
print(c.real)
print(c.imag)
```

```python
# Boolean => Binary
x = True
print(x)
print(type(x))  #bool

y = False
print(y)
print(type(y))  #bool
```

```python
# Built-IN Methods: 
    # Case-Conversion Methods:
        # .upper() => lowercase to uppercase
        # .lower() => uppercase to lowercase
        # .capitalize() => First letter Uppercase and rest Lowercase.
        # .title() => First Letter of each word is uppercase and rest lowercase
        # .swapcase() => uppercase to lowercase and vice versa


str3 = "this is beautiful."
print(str3.upper())
str4 = "MY NAME IS BOBBY"
print(str4.lower())
str5 = "i eaRnEd 5000 rS."
print(str5.capitalize())
print(str5.title())
print(str5.swapcase())
```

```python
        # is-methods:
            # .isdigit()  # all letters must be digits
            # .isalpha()
            # .isalnum()
            # .isspace()
str6 = "23456"
print(str6.isdigit())

str7 = "SDFjgf"
print(str7.isalpha())

str8 = "Apple500000"
print(str8.isalnum())

str9 = "    "
print(str9.isspace())

```
