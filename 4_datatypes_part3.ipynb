```python
# Tuple:
    # It is created ()
    # It is collection
    # it is immutable
    # it is heterogeneous
    # it allows duplicates
    # it is ordered
    # it supports indexing
```

```python
t1 = ()
print(t1)
print(type(t1))

t2 = (1, 2.2, 3_4j, False, [11, 22, 33], (100, "Hello", 2.2), "APPLE")
print(t2)
print(type(t2))
```
